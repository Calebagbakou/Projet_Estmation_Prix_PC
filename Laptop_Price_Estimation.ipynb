```python
#Feature Engineering
X = df1[['marque','titre', 'processeur', 'generation', 'ram_go', 'stockage_ssd',
       'stockage_hdd', 'carte_graphique', 'ecran', 'etat']]
y = df1["prix_fcfa"]
```

```python
#Identifier types de variables
X_num = X.select_dtypes(include=["int", "float"])
X_cat = X.select_dtypes(include=["object"])
X.dtypes
```

### Preprocessing

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
cat_cols = ["marque","processeur","generation","carte_graphique","ecran","etat"]
num_cols = ["ram_go", "stockage_hdd", "stockage_ssd"]

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), num_cols), 
        ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False), cat_cols) 
    ])
```

### Train/Test Split

```python
from sklearn.model_selection import train_test_split
X_train,X_test, y_train, y_test = train_test_split(X,y, test_size=0.15, random_state=42)
```

```python
#Application du preprocessing sur les données
X_train_p = preprocessor.fit_transform((X_train))
X_test_p  = preprocessor.transform(X_test)

```

```python
#  Modele 1 : Regression Lineaire 
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error, r2_score
reg = LinearRegression()
reg.fit(X_train_p, y_train)

y_pred = reg.predict(X_test_p)
r2 = r2_score(y_test,y_pred)
mae_lineaire = mean_absolute_error(y_test, y_pred)
rmse_lineaire = np.sqrt(mean_squared_error(y_test,y_pred))

print("Modele 1  Regression Lineaire ")
print(f"R^2 : {np.round(r2, 3)}")
print(f"MAE  : {np.round(mae_lineaire, 0)} FCFA")
print(f"RMSE : {np.round(rmse_lineaire, 0)} FCFA")
```

```python
# Modele 2  : Random Forest 
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor

pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestRegressor(n_estimators=300, max_depth=None, random_state=42, n_jobs=-1))
])

pipeline.fit(X_train, y_train)

pred_foret = pipeline.predict(X_test)
mae_foret = mean_absolute_error(y_test, pred_foret)
r2_foret = r2_score(y_test, pred_foret)

rmse_foret = np.sqrt(mean_squared_error(y_test,pred_foret))

print("Modele 2  Random Forest ")

print(f"R^2 : {np.round(r2_foret, 3)}")
print(f"RMSE (erreur moyenne absolue) : {np.round(rmse_foret, 0)} FCFA")
print(f"MAE (erreur moyenne absolue) : {np.round(mae_foret, 0)} FCFA")
```
